### `null` values detection and removal

P.S.: character_anime_works doesn't have null values

```python
favs = favs.dropna(subset=['username'])
favs
```

```python
character_nicknames = character_nicknames.dropna(subset=['nickname'])
character_nicknames
```

```python
characters = characters.dropna(subset=['name'])
characters
```

### Duplicate rows detection and removal

```python
print(f"Number of rows duplicated (favs): {favs.duplicated().sum()}\n")
print(f"Number of rows duplicated (character_anime_works): {character_anime_works.duplicated().sum()}\n")
print(f"Number of rows duplicated (character_nicknames): {character_nicknames.duplicated().sum()}\n")
print(f"Number of rows duplicated (characters): {characters.duplicated().sum()}")
```

```python
duplicati = characters[characters.duplicated(keep=False)]
duplicati
```

```python
favs = favs.drop_duplicates(keep="first")
character_anime_works = character_anime_works.drop_duplicates(keep="first")
character_nicknames = character_nicknames.drop_duplicates(keep="first")
characters = characters.drop_duplicates(keep="first")
```

```python
print(f"Number of rows duplicated (favs): {favs.duplicated().sum()}\n")
print(f"Number of rows duplicated (character_anime_works): {character_anime_works.duplicated().sum()}\n")
print(f"Number of rows duplicated (character_nicknames): {character_nicknames.duplicated().sum()}\n")
print(f"Number of rows duplicated (characters): {characters.duplicated().sum()}")
```

### Typo correction

```python
# 1. copy() --> used to modify onli the copy of the DataFrame created
favs = favs.copy()
characters = characters.copy()

# 2. Changing data types of columns with .loc[]
# That garantees that you are modifying the original DataFrame
characters.loc[:, 'character_mal_id'] = characters['character_mal_id'].astype('int64')
characters.loc[:, 'favorites'] = characters['favorites'].astype('int64')

# 3. Optimization Category for reduce RAM usage
favs.loc[:, 'fav_type'] = favs['fav_type'].astype('category')
character_anime_works.loc[:, 'role'] = character_anime_works['role'].astype('category')
```

### Saving in cleaned csv files

```python
readCsv.save_csv_file(favs, '../../cleaned_data/favs_cleaned.csv')
readCsv.save_csv_file(character_anime_works, '../../cleaned_data/character_anime_works_cleaned.csv')
readCsv.save_csv_file(character_nicknames, '../../cleaned_data/character_nicknames_cleaned.csv')
readCsv.save_csv_file(characters, '../../cleaned_data/characters_cleaned.csv')
```
