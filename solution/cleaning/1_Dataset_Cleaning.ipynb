### 1.1 Pulizia e Preparazione dei Dati
In questa fase verifichiamo la presenza di valori nulli, duplicati o anomalie strutturali.
Un dataset pulito è il prerequisito per conclusioni affidabili.
Identifichiamo inoltre il numero di utenti unici e la distribuzione iniziale delle categorie.

```python
import pandas as pd

# df = favs.merge(details, how='left', on='mal_id')
favs = pd.read_csv("../data/favs.csv")
character_anime_works = pd.read_csv("../data/character_anime_works.csv")
character_nicknames = pd.read_csv("../data/character_nicknames.csv")
characters = pd.read_csv("../data/characters.csv")
```

```python
# Drop rows non utili
favs = favs.dropna(subset=['username'])
favs
```

```python
# Eliminare doppioni su tutti i csv

# Per riga
#print(f"Numero di righe duplicate: {favs.duplicated().sum()}")
#print(f"Numero di righe duplicate: {character_anime_works.duplicated().sum()}")
#print(f"Numero di righe duplicate: {character_nicknames.duplicated().sum()}")
print(f"Numero di righe duplicate: {characters.duplicated().sum()}")
```

```python
# # keep=False mostra tutte le occorrenze del duplicato, non solo la seconda
duplicati = characters[characters.duplicated(keep=False)]
duplicati
```

```python
# Controllo duplicati su ID specifici
duplicati_id = characters.duplicated(subset=['character_mal_id']).sum()
print(f"ID duplicati: {duplicati_id}")
```

```python
characters = characters.drop_duplicates(subset=['character_mal_id'], keep="first")
characters
```

```python
character_nicknames = character_nicknames.dropna(subset=['nickname'])
character_nicknames
```

```python
characters = characters.dropna(subset=['name'])
characters
```
