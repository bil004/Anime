## 1.3 Cleaning and Data Preparation

In this phase we check for null values, duplicates, or structural anomalies to delete it and improve the dataset of `person_voice_works`, `profiles`, `recommendations` and `stats`.

---

### Loading Data from csv files

```python
from lib.dbconnection import create_db_engine
from lib import readSaveCsv as rsc
import pandas as pd

engine = create_db_engine()

person_voice_works = rsc.import_data(engine, "person_voice_works")
profiles = rsc.import_data(engine, "profiles")
recommendations = rsc.import_data(engine, "recommendations")
stats = rsc.import_data(engine, "stats")
```

### `null` values detection and removal

```python
profiles = profiles.dropna(subset=['username'])
```

### Duplicate rows detection and removal

```python
print(f"Number of rows duplicated (person_voice_works): {person_voice_works.duplicated().sum()}\\n")
print(f"Number of rows duplicated (profiles): {profiles.duplicated().sum()}\\n")
print(f"Number of rows duplicated (recommendations): {recommendations.duplicated().sum()}\\n")
print(f"Number of rows duplicated (stats): {stats.duplicated().sum()}")
```

```python
person_voice_works = person_voice_works.drop_duplicates(keep="first")
profiles = profiles.drop_duplicates(keep="first")
recommendations = recommendations.drop_duplicates(keep="first")
stats = stats.drop_duplicates(keep="first")

profiles['birthday_dt'] = pd.to_datetime(profiles['birthday'], errors='coerce')

profiles['birth_year'] = profiles['birthday_dt'].dt.year
profiles['age'] = 2025 - profiles['birth_year']

profiles = profiles.dropna(subset=['age', 'gender'])
profiles = profiles[profiles['gender'] != '']

print(f"Profili validi pronti per l'analisi: {len(profiles)}")
print(profiles[['username', 'age', 'gender']].head())
```

```python
print(f"Number of rows duplicated (person_voice_works): {person_voice_works.duplicated().sum()}\\n")
print(f"Number of rows duplicated (profiles): {profiles.duplicated().sum()}\\n")
print(f"Number of rows duplicated (recommendations): {recommendations.duplicated().sum()}\\n")
print(f"Number of rows duplicated (stats): {stats.duplicated().sum()}")
```

### Typo correction

```python
person_voice_works = person_voice_works.copy()
recommendations = recommendations.copy()

person_voice_works.loc[:, 'role'] = person_voice_works['role'].astype('category')
```

### Saving in cleaned csv files

```python
rsc.save_data(person_voice_works, 'person_voice_works', engine)
rsc.save_data(profiles, 'profiles', engine)
rsc.save_data(recommendations, 'recommendations', engine)
rsc.save_data(stats, 'stats', engine)
```
