## 1.3 Cleaning and Data Preparation

In this phase we check for null values, duplicates, or structural anomalies to delete it and improve the dataset of `person_voice_works`, `profiles`, `ratings`, `recommendations` and `stats`.

---

### Loading Data from csv files

```python
import pandas as pd

person_voice_works = pd.read_csv("../../data/person_voice_works.csv")
profiles = pd.read_csv("../../data/profiles.csv")
ratings = pd.read_csv("../../data/ratings.csv")
recommendations = pd.read_csv("../../data/recommendations.csv")
stats = pd.read_csv("../../data/stats.csv")
```

### `null` values detection and removal

```python
profiles = profiles.dropna(subset=['username'])
ratings = ratings.dropna(subset=['username'])
recommendations = recommendations.dropna(subset=['username'])
stats = stats.dropna(subset=['username'])
```

### Duplicate rows detection and removal

```python
print(f"Number of rows duplicated (person_voice_works): {person_voice_works.duplicated().sum()}\\n")
print(f"Number of rows duplicated (profiles): {profiles.duplicated().sum()}\\n")
print(f"Number of rows duplicated (ratings): {ratings.duplicated().sum()}\\n")
print(f"Number of rows duplicated (recommendations): {recommendations.duplicated().sum()}\\n")
print(f"Number of rows duplicated (stats): {stats.duplicated().sum()}")
```

```python
person_voice_works = person_voice_works.drop_duplicates(keep="first")
profiles = profiles.drop_duplicates(keep="first")
ratings = ratings.drop_duplicates(keep="first")
recommendations = recommendations.drop_duplicates(keep="first")
stats = stats.drop_duplicates(keep="first")
```

```python
print(f"Number of rows duplicated (person_voice_works): {person_voice_works.duplicated().sum()}\\n")
print(f"Number of rows duplicated (profiles): {profiles.duplicated().sum()}\\n")
print(f"Number of rows duplicated (ratings): {ratings.duplicated().sum()}\\n")
print(f"Number of rows duplicated (recommendations): {recommendations.duplicated().sum()}\\n")
print(f"Number of rows duplicated (stats): {stats.duplicated().sum()}")
```

### Typo correction

```python
person_voice_works = person_voice_works.copy()
profiles = profiles.copy()
ratings = ratings.copy()
recommendations = recommendations.copy()
stats = stats.copy()

person_voice_works.loc[:, 'role'] = person_voice_works['role'].astype('category')
profiles.loc[:, 'gender'] = profiles['gender'].astype('category')
recommendations.loc[:, 'recommendation_type'] = recommendations['recommendation_type'].astype('category')
stats.loc[:, 'status'] = stats['status'].astype('category')
```

### Saving in cleaned csv files

```python
import os
if not os.path.exists('../../cleaned_data'):
    os.makedirs('../../cleaned_data')

person_voice_works.to_csv('../../cleaned_data/person_voice_works_cleaned.csv', index=False)
profiles.to_csv('../../cleaned_data/profiles_cleaned.csv', index=False)
ratings.to_csv('../../cleaned_data/ratings_cleaned.csv', index=False)
recommendations.to_csv('../../cleaned_data/recommendations_cleaned.csv', index=False)
stats.to_csv('../../cleaned_data/stats_cleaned.csv', index=False)

```
