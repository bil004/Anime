# Top Ten anime analysis (Fans)

```python
import pandas as pd

favs = pd.read_csv("../../cleaned_data/favs_cleaned.csv")
details = pd.read_csv("../../cleaned_data/details_cleaned.csv")
```

```python
topAnime = pd.merge(details, favs, left_on="mal_id", right_on= "id")
topAnimeFans = topAnime.sort_values(by="favorites", ascending=False)
topAnimeFans = topAnimeFans.drop_duplicates(subset=['title'])
topAnimeFans
```
